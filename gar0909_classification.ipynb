<font color=yellow>讀取資料、分離群組資料、標準化資料及切割出訓練與測試資料</font>

```python
# load data/Wine.xlsx and separate features and labels and standardize the features

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

df = pd.read_excel('data/Wine.xlsx')
X = df.drop('Customer_Segment', axis=1)
y = df['Customer_Segment']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# split the data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3)
```

<font color=yellow>Logistic Regression</font>

<img src="pictures/Copilot_20260911_213253.png" alt="Multinomial Logistic Regression Diagram" width="600">


從最陽春的 Logistic Regression 開始，先將原始資料套用在訓練集上，並在測試集上評估模型的表現。


```python
# apply logistic regression on the training set and evaluate the model on the testing set

from sklearn.linear_model import LogisticRegression


LR_clf = LogisticRegression()
LR_clf.fit(X_train, y_train)
y_pred = LR_clf.predict(X_test)
print("Test Accuracy:", LR_clf.score(X_test, y_test))
```

<font color=yellow>SVM</font>

從最陽春的 SVM 開始，先將原始資料套用在訓練集上，並在測試集上評估模型的表現。

```python
# apply SVM on the training set and evaluate the model on the testing set
from sklearn.svm import SVC

SVM_clf = SVC()
SVM_clf.fit(X_train, y_train)
y_pred = SVM_clf.predict(X_test)
print("Test Accuracy:", SVM_clf.score(X_test, y_test))
```

<font color=yellow>Neural Network: 類別型輸出</font>

從最陽春的 Neural Network 開始，先將原始資料套用在訓練集上，並在測試集上評估模型的表現。

```python
# apply neural network on the training set and evaluate the model on the testing set
from sklearn.neural_network import MLPClassifier


NN_clf = MLPClassifier(hidden_layer_sizes=(20,), max_iter=1000, random_state=42)
NN_clf.fit(X_train, y_train)
y_pred = NN_clf.predict(X_test)
print("Test Accuracy:", NN_clf.score(X_test, y_test))
```

```python
# Do PCA on the standardized features
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
# apply logistic regression on the training set and evaluate the model on the testing set using PCA features

LR_clf_pca = LogisticRegression()
LR_clf_pca.fit(X_train_pca, y_train)
y_pred_pca = LR_clf_pca.predict(X_test_pca)
print("Test Accuracy (PCA):", LR_clf_pca.score(X_test_pca, y_test))
```

```python
# apply SVM on the training set and evaluate the model on the testing set using PCA features
SVM_clf_pca = SVC()
SVM_clf_pca.fit(X_train_pca, y_train)
y_pred_pca = SVM_clf_pca.predict(X_test_pca)
print("Test Accuracy (PCA):", SVM_clf_pca.score(X_test_pca, y_test))
```

```python
# apply neural network on the training set and evaluate the model on the testing set using PCA features
NN_clf_pca = MLPClassifier(hidden_layer_sizes=(20,), max_iter=1000, random_state=42)
NN_clf_pca.fit(X_train_pca, y_train)
y_pred_pca = NN_clf_pca.predict(X_test_pca)
print("Test Accuracy (PCA):", NN_clf_pca.score(X_test_pca, y_test))
```
