```python
# load a file from the data folder and print the first 5 rows

import pandas as pd

df2 = pd.read_excel("data/Wine.xlsx")
df2.head()
# take all columns except the last column
X = df2.iloc[:, :-1]
y = df2.iloc[:, -1] 
print(X.head())
print(y.head())
```

```python
# draw a boxplot of the data X

import matplotlib.pyplot as plt
# make a horizontal boxplot of the data X
# let the figure size be 10 by 5
plt.figure(figsize=(6, 3))
plt.boxplot(X, vert=False)
# add column names to the y-axis
plt.yticks(range(1, len(X.columns) + 1), X.columns)
plt.show()
```

```python
# standardize the data X using StandardScaler from sklearn.preprocessing

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# print(X_scaled)
# draw a boxplot of the standardized data X_scaled

import matplotlib.pyplot as plt
# make a horizontal boxplot of the data X
# let the figure size be 10 by 5
plt.figure(figsize=(6, 3))
plt.boxplot(X_scaled, vert=False)
# add column names to the y-axis
plt.yticks(range(1, len(X.columns) + 1), X.columns)
plt.show()
```
